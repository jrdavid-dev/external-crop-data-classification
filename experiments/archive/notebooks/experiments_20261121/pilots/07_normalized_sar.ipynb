# 07. Normalized Sentinel-1 features with optical EVI2

## Setup

```python
from pathlib import Path
import json
import sys

root = Path.cwd().resolve()
while not (root / "pyproject.toml").is_file():
    if root == root.parent:
        raise RuntimeError("Open this notebook inside the project repository")
    root = root.parent
if str(root / "src") not in sys.path:
    sys.path.insert(0, str(root / "src"))
if str(root) not in sys.path:
    sys.path.insert(0, str(root))  # experiments package

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from experiments.pilots import climate, clustering, data, earth_engine, evaluation, features
from experiments.tuning import tuning
pd.set_option("display.width", 220)
pd.set_option("display.max_columns", 40)
AREA = "pilots"


def show_and_save(figure, run, name):
    figure.tight_layout()
    figure.savefig(run / f"{name}.png", dpi=150)
    display(figure)
    plt.close(figure)


def repeated(name, X, rows, ok, seeds=(1, 2, 3), params=evaluation.PARAMS):
    """Mean and SD over random-forest seeds of every evaluation view."""
    summaries, details = [], []
    for seed in seeds:
        summary, holdout = evaluation.evaluate(name, X, rows, ok, params, seed)
        summaries.append({**summary, "seed": seed})
        details.extend({**r, "seed": seed} for r in holdout)
    return summaries, details


def summarize(summaries):
    table = pd.DataFrame(summaries)
    metrics = ["random_split_macro_f1", "spatial_cv_macro_f1", "dataset_holdout_macro_f1", "jecam_macro_f1",
               "jecam_carrot_f1", "jecam_cabbage_f1", "jecam_potato_f1"]
    grouped = table.groupby("feature_set", sort=False)
    out = grouped[metrics].agg(["mean", "std"]).round(3)
    out.insert(0, ("pixels_used", ""), grouped["pixels_used"].first())
    return out
```

## Inputs

```python
subset_run = ""  # Optional pipeline_outputs/pilots/subset_<id>; blank uses the latest
subset_run = Path(subset_run) if subset_run else tuning.latest_run(root, "subset", area=AREA)
rows, arrays = data.load(subset_run)
cube = features.stack(arrays["s2_values"])
valid = arrays["s2_valid"]
print("Subset:", subset_run, "| pixels:", len(rows))
```

## Radar variables

```python
sar = features.sar_variables(arrays["s1_values"])
sar_valid = np.isfinite(sar).all(axis=2)
zscored = features.temporal_zscore(sar[:, :, :2], sar_valid)
optical = features.optical_indices(arrays["s2_values"])
evi2_X, evi2_ok, evi2_cols = features.harmonic_features(optical[:, :, :1], valid, names=("EVI2",), form="amplitude")
optical_X, optical_ok, optical_cols = features.harmonic_features(optical, valid, names=features.OPTICAL_INDEX_NAMES, form="amplitude")
raw_X, raw_ok, raw_cols = features.harmonic_features(sar[:, :, :2], sar_valid, names=("VV", "VH"), form="full")
ratio_X, ratio_ok, ratio_cols = features.harmonic_features(sar[:, :, 2:], sar_valid, names=("VHVV", "CROSSPOL"), form="amplitude")
z_X, z_ok, z_cols = features.harmonic_features(zscored, sar_valid, names=("zVV", "zVH"), form="amplitude")
sets = {"s1_raw_dB_full": (raw_X, raw_ok, raw_cols),
        "s1_ratios_amplitude": (ratio_X, ratio_ok, ratio_cols),
        "s1_zscore_amplitude": (z_X, z_ok, z_cols),
        "evi2_amplitude": (evi2_X, evi2_ok, evi2_cols),
        "evi2_plus_s1_ratios": (np.hstack([evi2_X, ratio_X]), evi2_ok & ratio_ok, evi2_cols + ratio_cols),
        "optical_indices_plus_s1_ratios_zscore": (np.hstack([optical_X, ratio_X, z_X]), optical_ok & ratio_ok & z_ok,
                                                  optical_cols + ratio_cols + z_cols)}
common = np.logical_and.reduce([ok for _, ok, _ in sets.values()])
pd.DataFrame([{"feature_set": k, "predictors": X.shape[1], "pixels": int(ok.sum())} for k, (X, ok, _) in sets.items()])
```

## Evaluate on pixels usable by every set

```python
run = tuning.new_run(root, "normalized_sar", subset_run / "meta.json", area=AREA)
summaries, details = [], []
for name, (X, _, _) in sets.items():
    s, d = repeated(name, np.nan_to_num(X), rows, common)
    summaries += s
    details += d
    print(name, "done", flush=True)
tuning.save_records(run, "summary", summaries)
tuning.save_records(run, "dataset_holdout", details)
display(summarize(summaries))
```

## Spatial forward feature selection over normalized groups

```python
X, _, columns = sets["optical_indices_plus_s1_ratios_zscore"]
groups = features.column_groups(columns)
fast = {**evaluation.PARAMS, "n_estimators": 150}
selected, history = evaluation.forward_group_selection(np.nan_to_num(X), rows, common, groups, fast,
                                                       progress=lambda m: print(m, flush=True))
tuning.save_records(run, "selection_history", history)
display(pd.DataFrame(history).round(3))
chosen = [c for name in selected for c in groups[name]]
s, _ = repeated("selected_groups", np.nan_to_num(X[:, chosen]), rows, common)
summaries += s
tuning.save_records(run, "summary_with_selection", summaries)
display(summarize(summaries))
```

## Figure

```python
table = pd.DataFrame(summaries).groupby("feature_set", sort=False)[
    ["random_split_macro_f1", "spatial_cv_macro_f1", "dataset_holdout_macro_f1", "jecam_macro_f1"]].mean()
figure, axis = plt.subplots(figsize=(12, 4))
table.T.plot(kind="bar", ax=axis)
axis.set_ylim(0, 1)
axis.set_ylabel("Parcel macro-F1 (mean of seeds)")
axis.legend(fontsize=7)
show_and_save(figure, run, "normalized_sar")
print("Run folder:", run)
```
