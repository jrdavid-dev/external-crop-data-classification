# Aligned-matched validation with 20-day and 25-day onset gaps

Train two Random Forests with the baseline matched recipe. Both use the month-boundary and threshold-disagreement rules. The only timing limit changed between training sets is the maximum gap between the two observations that bracket green-up.

The notebook reads the saved preprocessing run and audits any newly completed development downloads. Fields already in the saved split keep their assignments. Newly downloaded fields are eligible for fitting according to their existing collection assignments. Both models score the identical original aligned-matched validation fields and pixels. This isolates the effect of changing the training data.

Run the cell with the project Python kernel. The notebook writes a separate training run with both models, parcel predictions, cohort membership and validation metrics. It uses local imagery and development labels.


```python
from pathlib import Path

from crop_pipeline.notebook_display import show_table
from crop_pipeline.training.aligned_gap_validation import train

root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").is_file())
run, rows = train(root)
print("Experiment outputs:", run)
score_columns = ("gap_days", "training_fields", "validation_fields", "macro_f1",
                 "carrot_f1", "cabbage_f1", "potato_f1")
show_table(
    [{key: row[key] for key in score_columns} for row in rows],
    title="Parcel F1 on the same locked validation fields",
)
show_table(
    [{key: value for key, value in row.items() if key == "gap_days" or key.startswith("new_")}
     for row in rows],
    title="Training fields from newly completed downloads",
)
```
